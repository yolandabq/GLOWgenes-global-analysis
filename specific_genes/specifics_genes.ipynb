# Genes específicos de panel

Vamos a considerar que los genes específicos de panel son aquellos que están en el top en muy pocos paneles y en el "bottom" la mayoría de las veces. 
Elegimos los threshold: 
- TOP: por encima del top 0.01
- BOTTOM: por debajo de 0.5

Ponemos los **thresholds**

```python
threshold_top=0.01 # por encima de
threshold_bottom=0.5 # por debajo de 
```

```python
max_valuedisease=gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"].max(axis=0) # el valor máximo de cada uno de los paneles. 
```

En base al valor máximo de cada panel, obtenemos cual va a ser la posición de corte del top y del bottom.

## Top

```python
division = [x*threshold_top for x in list(max_valuedisease)] 

max_valuedisease = pd.DataFrame(max_valuedisease)
max_valuedisease["top_ranking"] = division
max_valuedisease.T.iloc[1:]

comparison_df = pd.DataFrame([division] * gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"].shape[0], columns=max_valuedisease.index)
comparison_df.index = gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"].index
comparison_df # creamos este dataframe con el valor máximo del %top. Ahora compararemos nuestra tabla con esta (¿la posición 0,0 de glowmatrix es menor que la posición 0.0 de comparation_df?). Así calculamos el % de top. 
# estamos quitando los ceros. 
mask = (gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"] < comparison_df) & (gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"] != 0)

n_tops = mask.sum(axis = 1)
n_tops
```

```python
d = (mask.loc["SDHA"])

pd.set_option('display.max_rows', d.shape[0]+1)
print(d)
```

## Bottom

```python
division_bottom=[x*threshold_bottom for x in list(max_valuedisease[0])] 

max_valuedisease = pd.DataFrame(max_valuedisease)
max_valuedisease["bottom_ranking"] = division_bottom
max_valuedisease.T.iloc[1:]


comparison_bottom_df = pd.DataFrame([division_bottom] * gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"].shape[0], columns=max_valuedisease.index)
comparison_bottom_df.index = gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"].index
comparison_bottom_df # creamos este dataframe con el valor máximo del %top. Ahora compararemos nuestra tabla con esta (¿la posición 0,0 de glowmatrix es menor que la posición 0.0 de comparation_df?). Así calculamos el % de top. 

mask_bottom = (gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"] > comparison_bottom_df) & (gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"] != 0)

n_bottom = mask_bottom.sum(axis = 1)
n_bottom
```

# Genes específicos de categoría

Vamos a considerar que para que un gen sea específico de categoría, tiene que estar top en más del 50% de los paneles de su categoría y bottom en el resto. 

- De cada gen, obtenemos en qué paneles está en posición top (más arriba del 0.01) y en qué paneles está bottom (por debajo del 0.5). Habrá valores que quedarán en medio.
- Una vez tenemos esta información, vamos gen por gen, cogiendo los paneles pertenecientes a cada categoría. Si más de la mitad de los paneles son TOPs, consideramos que ese grupo es TOP para ese gen. Si más de la mitad de los paneles están BOTTOM, consideramos que ese panel es BOTTOM para ese gen. Si ni una cosa de la otra, es "MIDDLE".
- Seleccionamos los genes que tienen una única categoría TOP y tienen más de 10 BOTTOM (que sería más de la mitad de los paneles). 



```python
panel_class
```

```python
threshold_top=0.05 # por encima de
threshold_bottom=0.5 # por debajo de 
max_valuedisease=gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"].max(axis=0) # el valor máximo de cada uno de los paneles. 
```

```python
class_top_bottom = pd.DataFrame(columns=list(set(panel_class[0])), 
            index = gene_disease_matrix.index)
class_top_bottom
```

```python
division = [x*threshold_top for x in list(max_valuedisease)] 

max_valuedisease = pd.DataFrame(max_valuedisease)
max_valuedisease["top_ranking"] = division
max_valuedisease.T.iloc[1:]

comparison_df = pd.DataFrame([division] * gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"].shape[0], columns=max_valuedisease.index)
comparison_df.index = gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"].index
comparison_df # creamos este dataframe con el valor máximo del %top. Ahora compararemos nuestra tabla con esta (¿la posición 0,0 de glowmatrix es menor que la posición 0.0 de comparation_df?). Así calculamos el % de top. 

mask = (gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"] < comparison_df) & (gene_disease_matrix.loc[:, gene_disease_matrix.columns != "Cluster"] != 0)


top_bottom = mask
top_bottom
```

```python
for g in top_bottom.index:
    for c in set(panel_class[0]):
        position_class = top_bottom.loc[g,panel_class[panel_class[0]==c][1]]
        top_in_class = np.sum(position_class == "TOP")
        bottom_in_class = np.sum(position_class == "BOTTOM")
        n_panels_class = len(position_class)
        if top_in_class > n_panels_class*0.5:
            class_top_bottom.loc[g, c] = "TOP"
        elif bottom_in_class > n_panels_class*0.5:
            class_top_bottom.loc[g, c] = "BOTTOM"
        else: 
            class_top_bottom.loc[g, c] = "MIDDLE"
```

```python
gen_n_class_top = np.sum(class_top_bottom == "TOP", axis=1)
gen_n_class_bottom = np.sum(class_top_bottom == "BOTTOM", axis=1)
```

```python
ranking_n_class = pd.DataFrame(
    {'n_top': gen_n_class_top,
     'n_bottom': gen_n_class_bottom
    })

ranking_n_class
```
